## Cell 2: Sentence Word Frequency Counter
This cell defines a function that receives a sentence, cleans punctuation, converts words to lowercase, and counts word occurrences using Python's `collections.Counter`.

```python
from collections import Counter
import string

def count_word_frequency(sentence: str) -> dict[str, int]:
    """
    Reads a sentence and returns a dictionary of word frequencies.
    Strips punctuation and normalizes words to lowercase.
    """
    # Strip punctuation using str.translate
    translator = str.maketrans('', '', string.punctuation)
    clean_sentence = sentence.translate(translator)
    
    # Convert to lowercase and tokenize by whitespace
    words = [word.lower() for word in clean_sentence.split()]
    
    # Compute word counts
    frequency = Counter(words)
    return dict(frequency)

# Demonstration
sample_sentence = "AI engineers ship code in Python. Python is the foundation of modern AI, AI products, and AI workflows!"

print("--- Input Sentence ---")
print(sample_sentence)
print("\n--- Word Frequency Results ---")
counts = count_word_frequency(sample_sentence)
for word, count in sorted(counts.items(), key=lambda item: item[1], reverse=True):
    print(f"  '{word}': {count}")
```

## Cell 3: Text Cleaning & Normalization Pipeline
This cell defines a function that cleans messy input text by stripping punctuation, converting to lowercase, and collapsing multiple spaces/tabs/newlines into standard single spaces.

```python
import re
import string

def clean_text(input_text: str) -> str:
    """
    Normalizes and cleans input text:
    1. Lowercases text
    2. Strips punctuation
    3. Collapses irregular whitespace (spaces, tabs, newlines) into single spaces
    """
    # 1. Lowercase
    text = input_text.lower()
    
    # 2. Strip punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    # 3. Normalize whitespace
    cleaned = re.sub(r'\s+', ' ', text).strip()
    return cleaned

# Demonstration with messy input text
raw_text = """
  *** HELLO World!!!  Welcome   to  Day 2 of   AI Engineering... 
   Text preprocessing is CRITICAL for   LLM  Tokenization,   Prompting , and    RAG pipelines! 
"""

print("--- Raw Input Text ---")
print(repr(raw_text))
print("\n--- Cleaned & Normalized Text ---")
result = clean_text(raw_text)
print(result)
```
